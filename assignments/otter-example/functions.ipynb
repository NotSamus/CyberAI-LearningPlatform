**Question 1.** Write `square(x)` that returns x squared.

```python
def square(x):
    return x * x # SOLUTION
```

```python
square(2) == 4
```

```python
# HIDDEN
square(-3) == 9
```

**Question 2.** Write `total(numbers)` that returns the sum of a list.

```python
def total(numbers):
    # BEGIN SOLUTION
    result = 0
    for n in numbers:
        result += n
    return result
    # END SOLUTION
```

```python
total([1, 2, 3]) == 6
```

```python
# HIDDEN
total([]) == 0
```

# Hello GPU

First, check which device this notebook is running on.

```python
import torch

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", torch.cuda.get_device_name(0) if device == "cuda" else "CPU only")
```

## Train a tiny network

It learns `y = 2x + 1` from noisy points.

```python
x = torch.linspace(-1, 1, 2000, device=device).unsqueeze(1)
y = 2 * x + 1 + 0.1 * torch.randn_like(x)

model = torch.nn.Linear(1, 1).to(device)
opt = torch.optim.SGD(model.parameters(), lr=0.1)
for step in range(200):
    loss = torch.nn.functional.mse_loss(model(x), y)
    opt.zero_grad(); loss.backward(); opt.step()

print(f"weight={model.weight.item():.2f}  bias={model.bias.item():.2f}  loss={loss.item():.4f}")
```
